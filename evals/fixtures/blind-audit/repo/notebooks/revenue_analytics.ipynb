```python
subs = pd.read_sql(
    """
    SELECT s.id, c.email, c.name, s.plan_id, s.status, s.price_cents, s.quantity, p.interval
    FROM subscriptions s
    JOIN customers c ON c.id = s.customer_id
    JOIN plans p ON p.id = s.plan_id
    """,
    engine,
)
subs.head(3)
```

```python
active = subs[subs.status == "active"]
mrr_cents = (active.price_cents * active.quantity).sum()
print(f"MRR: ${mrr_cents / 100:,.2f}")
```
